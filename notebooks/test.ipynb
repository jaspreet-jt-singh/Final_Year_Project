```python
import yaml
from ultralytics import YOLO

model     = YOLO('G:/Projects/Final_Year_Project/models/yolo11s_indian.pt')
with open('G:/Projects/Final_Year_Project/notebooks/data.yaml', 'r', encoding='utf-8') as f:
    yaml_names = yaml.safe_load(f)['names']

print("ID  | data.yaml name     | model.names        | Match?")
print("-" * 55)
mismatches = []
for i, yaml_name in enumerate(yaml_names):
    model_name = model.names.get(i, 'MISSING')
    match      = '✅' if yaml_name == model_name else '❌'
    if yaml_name != model_name:
        mismatches.append(i)
    print(f"{i:2d}  | {yaml_name:<18s} | {model_name:<18s} | {match}")

print(f"\n{'✅ All labels match!' if not mismatches else f'❌ Mismatches at IDs: {mismatches}'}")
```

```python
from ultralytics import YOLO
from PIL import Image

model   = YOLO('G:/Projects/Final_Year_Project/models/yolo11s_indian.pt')
results = model.predict(
    source='G:/Projects/Final_Year_Project/data/indianfoodnet_yolo/test/images/00e616b729045e474b8838f5d4a257e8_jpg.rf.75bafa38ab7b778c9291ffff5c2263a5.jpg',
    conf=0.25,
    iou=0.45,
    imgsz=640,
    verbose=False
)

print(f"{'Class':<20} {'Confidence':>10} {'BBox (x1,y1,x2,y2)'}")
print("-" * 60)
for r in results:
    if len(r.boxes) == 0:
        print("⚠️  No detections — try lowering conf to 0.10")
    for box in r.boxes:
        cls_id   = int(box.cls)
        cls_name = model.names[cls_id]   # always use model.names, never hardcode
        conf     = float(box.conf)
        coords   = [round(x, 1) for x in box.xyxy[0].tolist()]
        print(f"{cls_name:<20} {conf:>10.2%} {coords}")
```
